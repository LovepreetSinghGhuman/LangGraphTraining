# Looping Graph


```python
from typing import Dict, TypedDict, List
from langgraph.graph import StateGraph, START, END 
import random
```

```python
class AgentState(TypedDict):
    name: str
    number: List[int]
    counter: int
```

```python
def greeting_node(state: AgentState) -> AgentState:
    """Initial greeting node."""
    state["name"] = f"Hello, {state['name']}! Let's start the computation."
    state["counter"] = 0
    return state

def random_number_node(state: AgentState) -> AgentState:
    """Generates a random number and appends it to the list."""
    state["number"].append(random.randint(1, 100))
    state["counter"] += 1
    return state
```
